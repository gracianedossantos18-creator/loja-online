**Soma igredientes **

```
#importaçõe
import sys
```

```
preco_cenoura = 4.5
preco_oleo = 12
preco_fermento = 15
preco_leite = 5
preco_acucar = 6
preco_ovos = 12
```

```
def soma_igredientes(tem_cenoura, tem_ovos, tem_oleo, tem_leite, tem_fermento, tem_acucar):
    total_compra = 0
    if tem_cenoura:
        total_compra = total_compra + preco_cenoura
    if tem_ovos:
        total_compra = total_compra + preco_ovos
    if tem_oleo:
        total_compra = total_compra + preco_oleo
    if tem_leite:
        total_compra = total_compra + preco_leite
    if tem_fermento:
        total_compra = total_compra + preco_fermento
    if tem_acucar:
        total_compra = total_compra + preco_acucar
    return total_compra
```

```
terminal_tem_cenoura = True
terminal_tem_ovo = True
terminal_tem_oleo = True
terminal_tem_leite = True
terminal_tem_fermento = True
terminal_tem_acucar = True

total = soma_igredientes(terminal_tem_cenoura, terminal_tem_ovo, terminal_tem_oleo, terminal_tem_leite, terminal_tem_fermento, terminal_tem_acucar)

print("Total dos igredientes: ", total)
```
